```python
# ==============================================
# BRONZE LAYER - Raw Data Ingestion
# ==============================================
# Purpose: Load all raw CSV files exactly as received
# No transformations, no cleaning, no business logic
# This layer preserves source data for traceability
# Architecture: Medallion (Bronze → Silver → Gold)
# Dataset: Olist Brazilian E-Commerce (99k orders)
# ==============================================
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Ecommerce-Bronze-Layer") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

print("Spark Session Started Successfully")
print(f"Spark Version: {spark.version}")
```

```python
import os

# Define source and destination paths
RAW_DATA_PATH = "data/"
BRONZE_PATH = "bronze/"

# Create bronze directory if it doesn't exist
os.makedirs(BRONZE_PATH, exist_ok=True)

# Define all source files with meaningful names
source_files = {
    "orders"          : "olist_orders_dataset.csv",
    "order_items"     : "olist_order_items_dataset.csv",
    "customers"       : "olist_customers_dataset.csv",
    "products"        : "olist_products_dataset.csv",
    "sellers"         : "olist_sellers_dataset.csv",
    "order_payments"  : "olist_order_payments_dataset.csv",
    "order_reviews"   : "olist_order_reviews_dataset.csv",
    "geolocation"     : "olist_geolocation_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

print("Starting Bronze Layer Ingestion...")
print("=" * 45)
```

```python
# Track ingestion results
ingestion_summary = []

for table_name, file_name in source_files.items():
    
    source_path = RAW_DATA_PATH + file_name
    bronze_path = BRONZE_PATH + table_name
    
    # Read CSV exactly as-is — no modifications
    df = spark.read.csv(
        source_path,
        header=True,
        inferSchema=True
    )
    
    row_count = df.count()
    col_count = len(df.columns)
    
    # Write to Parquet — this is what triggers Spark to actually read the file
    df.write.mode("overwrite").parquet(bronze_path)
    
    ingestion_summary.append({
        "table"   : table_name,
        "rows"    : row_count,
        "columns" : col_count,
        "status"  : "SUCCESS"
    })
    
    print(f"✓ {table_name:<25} {row_count:>7} rows  |  {col_count} columns")

print("=" * 45)
print("Bronze ingestion complete.")
```

```python
import os

print("Bronze layer files saved:")
print("=" * 45)
for folder in os.listdir("bronze/"):
    path = f"bronze/{folder}"
    files = os.listdir(path)
    parquet_files = [f for f in files if f.endswith(".parquet")]
    print(f"✓ {folder:<25} {len(parquet_files)} parquet file(s)")
```
